7.1. Nối nhiều DataFrame


```python

# Tạo 2 DataFrame mẫu
df1 = pd.DataFrame({'Name':['John', 'Smith', 'Paul'], 'Age':[25, 30, 50]}, index=[0, 1, 2])
df2 = pd.DataFrame({'Name':['Adam', 'Smith'], 'Age':[26, 11]}, index=[3, 4])

# 1. Nối 2 DataFrame (theo chiều dọc)
df_concat = pd.concat([df1, df2])
print('Concate two DataFrame\n', df_concat)

# 2. Xóa các hàng có giá trị trùng nhau ở cột Name
# Lưu ý: Smith xuất hiện ở cả df1 và df2, lệnh này sẽ xóa bớt 1 Smith
print('\nDrop duplicate values in DataFrame\n', df_concat.drop_duplicates('Name'))

# 3. Sắp xếp bảng theo giá trị cột Age (từ nhỏ đến lớn)
print('\nSort values in DataFrame\n', df_concat.sort_values('Age'))

# 4. Đổi tên cột
df_rename = df_concat.rename(columns={'Name': 'Surname', 'Age': 'Age_pupil'})
print('\nRename columns of DataFrame\n', df_rename)
```

7.3. Gom nhóm dữ liệu DataFrame

```python
import pandas as pd
import numpy as np

# Tạo dữ liệu giả lập về các loài cá
data = {
    'Species': ['Bream', 'Bream', 'Roach', 'Roach', 'Pike', 'Pike', 'Bream'],
    'Weight': [242, 290, 120, 150, 500, 600, 340],
    'Height': [11.5, 12.4, 6.2, 7.1, 15.0, 16.5, 13.0]
}
df_csv = pd.DataFrame(data)

# 1. Tính trung bình (mean) cho tất cả các cột số theo từng loài cá
print("Trung bình theo loài:\n", df_csv.groupby(['Species']).mean())

# 2. Tìm trọng lượng nhỏ nhất (min) của từng loài cá
print("\nTrọng lượng nhỏ nhất theo loài:\n", df_csv.groupby(['Species'])['Weight'].min())

# 3. Gom nhóm theo nhiều cột (Loài và Chiều cao) rồi tìm trọng lượng nhỏ nhất
print("\nMin Weight theo Species và Height:\n", df_csv.groupby(['Species', 'Height'])['Weight'].min())
```

3.1. Tensor 0-chiều

```python
import tensorflow as tf

# Tạo một Tensor hạng 0 (Rank 0 Tensor)
# Đây thực chất là một hằng số (Scalar)
rank_0_tensor = tf.constant(4)

print(rank_0_tensor)
```

```python
import tensorflow as tf

# 1. Mặc định là số nguyên 32-bit (int32)
rank_0_tensor = tf.constant(4)
print(rank_0_tensor)

# 2. Ép kiểu về số nguyên 16-bit (int16) - tốn ít bộ nhớ hơn
r1 = tf.constant(9, tf.int16)
print(r1)

# 3. Tạo Tensor số thực (float32) - dùng cho các phép toán có dấu phẩy
r1_float = tf.constant(1.12, tf.float32)
print(r1_float)

# 4. Tạo Tensor kiểu chuỗi văn bản (string)
r1_string = tf.constant('Hello Vanlang University', tf.string)
print(r1_string)
```

3.2. Tensor 1-chiều

```python

# Tạo một Rank 1 Tensor (Vector) từ một danh sách số thực
rank_1_tensor = tf.constant([2.0, 3.0, 4.0])

print(rank_1_tensor)
```

3.3. Tensor 2-chiều (ma trận)

```python
rank_2_tensor = tf.constant([[1,2],
                            [3,4],
                            [5,6]], dtype=tf.float16

)
print(rank_2_tensor)
```
